**Тапсырма 38:** Сандар тізімін (list) қабылдап, сол сандардың орташа мәнін табатын функция жазыңыз.

```
def calculate_average(list):
    sum = 0
    for num in list:
        sum += num
    average = sum / len(list)
    return average

list = [2, 4, 6, 8, 10]
average = calculate_average(list)
print("The average of the list is:", average)
```

**Тапсырма 39:** `*args` қолдану: Кез келген сандағы аргументтерді қабылдап, олардың барлығының қосындысын қайтаратын функция жазыңыз.

```
# Кодыңызды осында жазыңыз
def sum_all(*args):
    total = 0
    for num in args:
        total += num
    return total
sum_all(1,2,3,4,5,6,7)
```

**Тапсырма 40:** Бір функциядан екі мән қайтару (Tuple): Тіктөртбұрыштың ұзындығы мен ені берілгенде, оның периметрі мен ауданын бірдей қайтаратын функция жазыңыз.

```
# Кодыңызды осында жазыңыз
def rectangle_properties(length, width):
    perimeter = 2 * (length + width)
    area = length * width
    return perimeter, area  # tuple ретінде қайтарылады
rectangle_properties(5, 3)
```

**Тапсырма 41 (Калькулятор):** Екі сан және математикалық оператор (`+`, `-`, `*`, `/`) қабылдап, нәтижені қайтаратын функция жазыңыз. Қате оператор енгізілсе, ескерту жасаңыз.

```
# Кодыңызды осында жазыңыз
def calculate(a, b, operator):
    if operator == "+":
        return a + b
    elif operator == "-":
        return a - b
    elif operator == "*":
        return a * b
    elif operator == "/":
        if b == 0:
            return "Қате: 0-ге бөлуге болмайды"
        return a / b
    else:
        return "Қате: белгісіз оператор"
calculate(3,8,'+')
```

```
# Кодыңызды осында жазыңыз
def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32
celsius_to_fahrenheit(28)
```

**Тапсырма 43 (Дауысты дыбыстар):** Мәтін берілген. Оның ішінде неше дауысты әріп (а, е, и, о, у, т.б.) бар екенін санайтын функция құрыңыз.

```
# Кодыңызды осында жазыңыз
def count_vowels(text):
    vowels = "аеёиоуыіэюяәөұүиAEЁИОУЫІЭЮЯӘӨҰҮИ"
    count = 0

    for char in text:
        if char in vowels:
            count += 1

    return count
count_vowels('қанағаттандырымағандықтарыңыздан')
```

**Тапсырма 49:** Фибоначчи қатарының алғашқы `n` мүшесін (0, 1, 1, 2, 3, 5, 8...) шығаратын функция жазыңыз.

```
# Кодыңызды осында жазыңыз
def fib(n):
    if n <= 1:
        return n
    else:
        return fib(n-1) + fib(n-2)

# Print first 10 terms
n = 10
if n > 0:
    for i in range(n):
        print(fib(i), end=" ")
```

```
# import random
# Кодыңызды осында жазыңыз

import random
def password():
  char = 'abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789'
  pwd = ''
  for i in range(8):
    pwd += random.choice(char)
  return pwd
print(password())
```
